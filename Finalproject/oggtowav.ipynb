```python
import os
import librosa
import soundfile as sf
from multiprocessing import Pool, cpu_count

# ====== PATHS ======
input_dir = "/home/users/ss1482/Final project/birdtrainold"
output_dir = "/home/users/ss1482/Final project/birdtrain_wav"

os.makedirs(output_dir, exist_ok=True)

# ====== SETTINGS ======
TARGET_SR = 32000  # BirdCLEF standard

# ====== GET FILES ======
files = [f for f in os.listdir(input_dir) if f.lower().endswith(".ogg")]
print(f"Found {len(files)} files.")

# ====== CONVERSION FUNCTION ======
def convert_file(filename):
    input_path = os.path.join(input_dir, filename)
    output_path = os.path.join(output_dir, filename.rsplit(".", 1)[0] + ".wav")

    try:
        audio, sr = librosa.load(input_path, sr=TARGET_SR, mono=True)
        sf.write(output_path, audio, TARGET_SR)
        return f"Done: {filename}"
    except Exception as e:
        return f"Failed: {filename} | {e}"

# ====== MULTIPROCESSING ======
workers = min(8, cpu_count())  # adjust if needed
print(f"Using {workers} workers...")

with Pool(workers) as pool:
    for i, msg in enumerate(pool.imap_unordered(convert_file, files)):
        if i % 100 == 0:
            print(f"{i}/{len(files)} processed")

print("Conversion complete.")
print(f"Saved to: {output_dir}")
```
