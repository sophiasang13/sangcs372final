```python
import sys, os, torch

repo_path = "/home/users/ss1482/sangcs372final/audioset_tagging_cnn"
sys.path.append(repo_path)
sys.path.append(os.path.join(repo_path, "pytorch"))

from pytorch.models import Cnn14

device = 'cuda' if torch.cuda.is_available() else 'cpu'

model = Cnn14(
    sample_rate=32000,
    window_size=1024,
    hop_size=320,
    mel_bins=64,
    fmin=50,
    fmax=14000,
    classes_num=527
)

checkpoint = torch.load(
    os.path.join(repo_path, "Cnn14.pth"),  # adjust if needed
    map_location=device
)

model.load_state_dict(checkpoint['model'])
model.to(device)
model.eval()

print("Model ready")
```
